##   عملیات ریاضی و متن

###  عملیات ریاضی
- **جمع**: `+`
- **تفریق**: `-`
- **ضرب**: `*`
- **تقسیم**: `/`
- **تقسیم صحیح**: `//`
- **توان**: `**`
- **باقی‌مانده**: `%`

###  عملیات متنی
- **اتصال متن**: استفاده از `+`
- **تکرار**: استفاده از `*`
- **طول**: استفاده از `len()`

```python
4 + 5 #جمع
```

```python
6 - 8 #کم(تفریق)
```

```python
9 - 4 #کم(تفریق)
```

```python
4.5 * 60 #ضرب 
```

```python
4 ** 2 #توان
```

```python
3 / 2 #تقسیم
```

```python
3 // 2 #تقسیم صحیح
```

```python
3 % 2 #باقیمانده
```

```python
number_1 = 3
number_2 = 10
print(number_1 + number_2)
print(number_1 - number_2)
print(number_1 * number_2)
print(number_1 / number_2)
print(number_1 // number_2)
print(number_1 % number_2)
print(number_1 ** number_2)
```

```python
# مثال
number_of_cell_1 = 2 #تعداد سلول 
number_of_cell_2 = 7 #تعداد سلول 
total_number_of_cell_1_2 = number_of_cell_1 + number_of_cell_2 #مجموع سلول ها 
print("cell_1 + cell_2:",total_number_of_cell_1_2)
```

```python
number_of_cell_2 = 19 #تعداد سلول 
number_of_cell_3 = 6 #تعداد سلول 
minus_number_of_cell_2_3 = number_of_cell_2 - number_of_cell_3 #مجموع سلول ها 
print("cell_2 - cell_3:",minus_number_of_cell_2_3)
```

```python
# ۲.۳ مثال عملی: عملیات ریاضی در زیست‌شناسی

# تقسیم سلول‌ها
initial_cells = 1 #تعداد سلول 
division_time_minutes = 20 #زمان تقسیم سلولی
total_time_hours = 2 # کل زمان برای تقسیم 
total_time_minutes = total_time_hours * 60 # تبدیل ساعت به دقیقه 

# تعداد تقسیم‌ها
number_of_divisions = total_time_minutes // division_time_minutes # تعداد تقسیم در زمان دو ساعت 
print(f"در {total_time_hours} ساعت {number_of_divisions} تقسیم سلولی اتفاق می‌افتد") # نمایش تعداد تقسم سلولی 

# تعداد سلول‌های نهایی
final_cells = initial_cells * (2 ** number_of_divisions) # تعداد کل سلول های بعد از اتمام تقسیم
print(f"از ۱ سلول شروع کنیم، بعد از {total_time_hours} ساعت: {final_cells} سلول") # نمایش تعداد کل سلول ها بعد تقسیم 
```

## عملیات متنی

```python
# DNA sequence
dna_sequence = "ATGC"  # توالی دی ان ای
print(f"دنباله DNA: {dna_sequence}")# نمایش دی ان ای 
print(f"طول دنباله: {len(dna_sequence)}")# نمایش طول دی ان ای
```

```python
# تکرار DNA
repeated_dna = dna_sequence * 3 #تکرار در دی ان ای 
print(f"اگر ۳ بار تکرار شود: {repeated_dna}")# نمایش دی ان ای 
```

```python
# اتصال رشته‌های DNA
gene1 = "ATGATTCGTACGGGTTTACG" # ژن اول
gene2 = "TAAATTCGGGGCATTTACGG" # ژن دوم 
combined_gene = gene1 + gene2 # اتصال دو ژن 
print(f"ترکیب دو ژن: {combined_gene}") # نمایش دو ژن 
```

```python
dna = "ACGAAATTTCGGGGCATAACAAGCACGAAATTTCGGGGCATAACAAGCACGAAATTTCGGGGCATAACAAGC" # دی ان ای  
print("DNA :", dna) # پرینت دی ان ای 
rna = dna.replace("T", "U")#تبدیل دی ان ای به ار ان ای 
print("RNA :", rna) #پرینت ار ان ای
length = len(dna) #طول دی ان ای 
print("DNA length :", length) #پرینت طول دی ان ای 
gc_content = dna.count("C") + dna.count("G") #تعداد بازهای گوانین و سیتوزین
print("G and C content :",gc_content) #پرینت تعداد کل  گوانین و سیتوزین 
at_content = dna.count("A") + dna.count("T") #تعداد بازهای آدنین و تیمین 
print("A and T content :",at_content) #پرینت تعداد کل بازهای آدنین و تیمین
gc_percentage = gc_content / length * 100
print("G and C percentage :", gc_percentage)
at_percentage = at_content /length * 100
print("A and T percentage :", at_percentage)
```
