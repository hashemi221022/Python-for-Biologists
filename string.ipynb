# String

## معرفی رشته‌ها در پایتون
این نوت‌بوک به بررسی مبانی رشته (`string`) در پایتون می‌پردازد، از جمله دسترسی به حروف، تکه‌برداری و برش‌های مختلف. 
مثال‌ها به شما کمک می‌کنند تا مفهوم اندیس‌دهی مثبت و منفی و همچنین گام‌های برش (`step`) را درک کنید.

```python
string = "salam"  # تعریف رشته اولیه با مقدار salam
```

```python
print(string)  # چاپ کل رشته
```

```python
string[0]  # اولین حرف رشته (اندیس 0)
```

```python
string[1]  # حرف دوم رشته (اندیس 1)
```

```python
string[4]  # حرف پنجم رشته (اندیس 4)
```

```python
string[1:3]  # زیررشته از اندیس 1 تا قبل از اندیس 3
```

```python
string[3:4]  # زیررشته شامل حرف چهارم تا قبل از حرف پنجم
```

```python
string[-1]  # آخرین حرف رشته با استفاده از اندیس منفی
```

```python
string[-2]  # دومین حرف از انتهای رشته
```

```python
string[-3]  # سومین حرف از انتهای رشته
```

```python
string[:]  # کل رشته را به صورت یک کپی باز می‌گرداند
```

```python
string[:3]  # زیررشته از ابتدا تا قبل از اندیس 3
```

```python
string[::2]  # هر دو حرف یکبار انتخاب می‌شود (گام 2)
```

```python
string[::-1]  # رشته را معکوس می‌کند
```

## نمونه‌های برش رشته
در این بخش، تفاوت بین `[:]` و `[::]` و دیگر شکل‌های برش را بررسی می‌کنیم. 
با استفاده از `course_name` نشان می‌دهیم چگونه می‌توان زیررشته‌ای از ابتدای رشته یا با گام مشخص برداشت.

```python
course_name = "biotechnology"  # تعریف رشته جدید برای نمونه‌های برش
```

```python
course_name[:]  # هم رشته را کپی می‌کند و هم معادل رشته اصلی است
```

```python
course_name[::]  # همان `[:]` است؛ زیررشته کامل با گام پیش‌فرض 1
```

```python
course_name[0:13:1]  # از حرف اول تا حرف آخر با گام 1
```

```python
course_name[0:13:2]  # زیررشته با گام 2؛ هر دو حرف یکبار انتخاب می‌شود
```

```python
course_name[0:13:3]  # زیررشته با گام 3؛ هر سه حرف یکبار انتخاب می‌شود
```

```python
course_name[0:13:4]  # زیررشته با گام 4
```

```python
course_name[0:13:5]  # زیررشته با گام 5
```

```python
course_name[0:13:6]  # زیررشته با گام 6
```
